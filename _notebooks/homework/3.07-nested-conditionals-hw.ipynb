### Popcorn Hack 1: Test the after-school hw plan

```python
is_weekday = True
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
```

##### Printed Message: Finish homework

##### Branches Followed: Outer branch = is_weekday (True), Inner branch = else (because homework_finished is False).

##### Values to Reach Other Outcomes:

##### To print "Free time": Set is_weekday = True and homework_finished = True.

##### To print "Weekend plan": Set is_weekday = False.

### Popcorn Hack 2: Repair Student Account Check

##### Why the original code was not nested:

##### The if has_permission: check was placed at the exact same indentation level as if logged_in:.

##### This caused the permission check to run as a completely separate statement instead of waiting for logged_in to be evaluated first.

```python
logged_in = True
has_permission = False

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
```

##### Tested Outcomes & Messages:

##### Outcome 1 (logged_in = True, has_permission = True): Prints Account found and Page opened.

##### Outcome 2 (logged_in = True, has_permission = False): Prints Account found and Permission needed.

##### Outcome 3 (logged_in = False): Prints Log in first.

### Popcorn Hack 3: Lunch and Sports Practice

##### Execution Order Explanation:

##### Python checks the outer condition (lunch_packed) first before looking at any inner code.

##### The inner condition (practice_after_school) is only evaluated after the outer path is selected.

```python
lunch_packed = True
practice_after_school = True

if lunch_packed:
    if practice_after_school:
        print("Packed lunch with practice - bring an extra snack!")
    else:
        print("Packed lunch without practice - normal day.")
else:
    if practice_after_school:
        print("Buying lunch with practice - buy extra food!")
    else:
        print("Buying lunch without practice - buy normal lunch.")
```
