### Homework Hack

```python
# Game settings and profile
game_title = "UESL Star Trail"
reduced_motion = True

# Initial level scores
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

# Calculate initial total
total_stars = level_one_stars + level_two_stars + level_three_stars

# 1. Print initial state
print(game_title)
print("Reduced Motion:", reduced_motion)
print("Initial Total Stars:", total_stars)

# 2. Add bonus of 5 to level two
level_two_stars = level_two_stars + 5

# Print stored total (unupdated)
print("Stored Total (Before Recalculation):", total_stars)

# Recalculate total
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Updated Total (After Recalculation):", total_stars)
```

### Test Case 1

##### Prediction: Initial total = 40, Stored total = 40, Recalculated total = 45

##### Actual: 40,40,45
##### Explanation: total_stars remained 40 after adding the bonus until line 21 explicitly recomputed the sum.

### Test Case 2

##### Prediction: Initial total = 40, Stored total = 40, Recalculated total = 40

##### Actual: 40,40,40
##### Adding 0 to level_two_stars keeps all totals unchanged at 40.

### Design Thinking note:
##### Player need: motion sensitive players need reduced fast visual animations to play comfortables
##### Goal: Store player accessibility preferences and progress accordingly
##### Representation: bool for reduced_motion (true/false) and seperate int variables for level scores
##### Prototype: Python scripts initializing player options + score recalculations
##### Test/Revision: Verified that recalculating total_stars is required whenever any individual level score updates.
##### Why redused motion should be a player choice: Accessibility settings should always be user-configurable rather than forced, ensuring players with intellectual, developmental, or sensory needs can customize their gameplay experience safely.
