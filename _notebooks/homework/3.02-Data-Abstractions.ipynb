### Homework Hack

```python
# Initial level collection: each element represents stars collected for a level in order
level_stars = [12, 18, 10]

# Add 5 to the second level (index 1)
level_stars[1] = level_stars[1] + 5

# Append a new level with 7 stars
level_stars.append(7)

# Calculate total using a list traversal loop
total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print("Level Collection:", level_stars)
print("Level Count:", len(level_stars))
print("Total Stars:", total_stars)
```

### Test Case 2

```python
level_stars = []

# Skip second-level update if index 1 does not exist
if len(level_stars) > 1:
    level_stars[1] = level_stars[1] + 5

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print("Level Count:", len(level_stars))
print("Total Stars:", total_stars)
```

##### Prediction: Count: 0, Total: 0
##### Actual: Level Count: 0, Total Stars: 0
##### Explanation: The for loop executes zero times when given an empty list, demonstrating that data abstraction avoids hardcoded index errors.


Managing Complexity Explanation:
Without a list, managing 100 levels would require defining 100 individual variables (level_1_stars through level_100_stars) and writing a massive addition expression (level_1_stars + level_2_stars + ...). Using a list data abstraction lets us store all levels in one variable and use a single 2-line for loop to total any number of levels automatically.

### Design Thinking
Player/Maker Need: Game creators need to easily add, remove, or reorder game levels without rewriting score-tracking code.

Goal: Use a list data abstraction to handle variable level counts smoothly.

Representation Considered: Individual integer variables vs. a single Python list. The list was selected because it maintains level sequence ordering and supports dynamic expansion.

Prototype: Python script utilizing append() for new levels and a for loop traversal for dynamic summation.

Test/Revision: Tested the empty list [] edge case and added an explicit length check (len(level_stars) > 1) before accessing index 1 to prevent IndexError exceptions.

Level Label Recommendation: Provide human-readable display names in the UI (e.g., "Level 1: Forest", "Level 2: Cavern") so players don't have to understand zero-based computer indexing.
