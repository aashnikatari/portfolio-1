**Fix:** Line 5 in the array needed ``.

```python
// 1. Get references to the interactive elements
const toggleTheme = document.getElementById("toggleTheme");
const toggleImage = document.getElementById("toggleImage");
const toggleAlert = document.getElementById("toggleAlert");

const activeCount = document.getElementById("activeCount");
const alertBox = document.getElementById("alertBox");
const imageBox = document.getElementById("imageBox");

const allToggles = document.querySelectorAll(".ocs__toggle-input");

// 2. Update the active toggle counter
// Insight: Querying all checkboxes lets us loop through them
// dynamically using forEach to calculate the total state.
function updateCounter() {
    let count = 0;

    allToggles.forEach(function(t) {
        if (t.checked) {
            count++;
        }
    });

    activeCount.textContent = count;
}

// 3. Alert toggle
// Insight: The .checked property is a Boolean value (true/false)
// that directly controls the .hidden attribute of DOM elements.
toggleAlert.addEventListener("change", function() {
    alertBox.hidden = !toggleAlert.checked;
    updateCounter();
});

// 4. Image toggle
// Insight: Using a change listener makes the update happen
// immediately when the user changes the checkbox.
toggleImage.addEventListener("change", function() {
    imageBox.hidden = !toggleImage.checked;
    updateCounter();
});

// 5. Dark mode toggle
toggleTheme.addEventListener("change", function() {
    updateCounter();
});

// 6. Set the initial counter
updateCounter();
```
