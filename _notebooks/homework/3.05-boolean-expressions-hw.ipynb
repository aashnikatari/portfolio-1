### Homework Hack

```python
# Homework Hack – SFI Backend Validator

existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006"
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026"
    }
]

# Build and test your validator below:

def validate_record(record):
    # Check 1: non-empty product name
    has_name = bool(record.get("product_name"))
    
    # Check 2: accepted racing category
    valid_cat = record.get("category") in valid_categories
    
    # Check 3: non-empty spec number
    has_spec = bool(record.get("spec_number"))
    
    # Check 4: effective date exists
    has_date = bool(record.get("effective_date"))
    
    # Check 5: spec number does NOT already exist in backend database
    unique_spec = record.get("spec_number") not in existing_spec_numbers
    
    # Evaluate all conditions together
    is_valid = has_name and valid_cat and has_spec and has_date and unique_spec
    return is_valid

# Run tests and print results for each candidate record
for idx, record in enumerate(test_records, 1):
    status = "ACCEPTED" if validate_record(record) else "REJECTED"
    print(f"Record #{idx} ({record['product_name']}): {status}")
```
